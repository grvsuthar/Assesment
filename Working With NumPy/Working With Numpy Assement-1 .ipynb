```python
import pandas as pd
```

```python
url = "https://github.com/TopsCode/Data_Analysis_2024/blob/main/ALL_CSV/banklist.csv?raw=true"
banks = pd.read_csv(url)
```

```python
print("Head of the DataFrame:")
print(banks.head())
```

```python
print("\nColumn Names:")
print(banks.columns.tolist())
```

```python
# 4. Count how many unique states (ST) are represented
print("\nNumber of unique states (ST):")
print(banks['ST'].nunique())
```

```python
# 5. Get a list of all the unique states
print("\nList of all states:")
print(banks['ST'].unique())
```

```python
# 6. Top 5 states with the most failed banks
print("\nTop 5 states with the most failed banks:")
print(banks['ST'].value_counts().head(5))
```

```python
# 7. Top 5 acquiring institutions
print("\nTop 5 acquiring institutions:")
print(banks['Acquiring Institution'].value_counts().head(5))
```

```python
# 8. State Bank of Texas acquisitions
sbt = banks[banks['Acquiring Institution'] == 'State Bank of Texas']
count_sbt_total = sbt.shape[0]
count_sbt_tx = sbt[sbt['ST'] == 'TX'].shape[0]

print(f"\nState Bank of Texas has acquired {count_sbt_total} banks.")
print(f"Out of them, {count_sbt_tx} were in Texas.")
```

```python

# 9. Most common city for bank failure in California
ca_banks = banks[banks['ST'] == 'CA']
common_city = ca_banks['City'].value_counts().idxmax()
print(f"\nMost common city in California for bank failure: {common_city}")
```
