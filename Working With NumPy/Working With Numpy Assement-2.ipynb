```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
url = "https://github.com/TopsCode/Data_Analysis_2024/blob/main/ALL_CSV/historical_automobile_sales.csv?raw=true"
df = pd.read_csv(url)
```

```python
df['Year'] = df['Year'].astype(int)
```

```python
# Q1: Line chart for total automobile sales per year
sales_by_year = df.groupby('Year')['Automobile_Sales'].sum()
sales_by_year.plot(kind='line', marker='o', title='Total Automobile Sales Over Years')
plt.xlabel('Year')
plt.ylabel('Total Sales')
plt.grid(True)
plt.show()
```

```python
# Q2: Line chart for each vehicle type over time
plt.figure(figsize=(10,6))
for vehicle in df['Vehicle_Type'].unique():
    subset = df[df['Vehicle_Type'] == vehicle]
    sales_per_year = subset.groupby('Year')['Automobile_Sales'].sum()
    plt.plot(sales_per_year, marker='o', label=vehicle)

plt.title("Automobile Sales Trends by Vehicle Type")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Q3: Seaborn plot: Compare recession vs non-recession sales per vehicle type
plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Vehicle_Type', y='Automobile_Sales', hue='Recession')
plt.title("Sales per Vehicle Type: Recession vs Non-Recession")
plt.xticks(rotation=45)
plt.show()
```

```python
# Q4: Comparison of sales by vehicle type and recession
comparison = df.groupby(['Vehicle_Type', 'Recession'])['Automobile_Sales'].sum().unstack()
comparison.plot(kind='bar', title='Sales Comparison: Recession vs Non-Recession')
plt.xlabel('Vehicle Type')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```
